## 0. Setup

Sesuaikan `DATA_FILE` dengan nama file yang sudah kamu unduh ke `data/raw/`.


```python
from pathlib import Path

import duckdb
import polars as pl

from src.data_loading import hitung_missing, resolve_data_path, ringkasan_dataframe

print(f"Polars : {pl.__version__}")
print(f"DuckDB : {duckdb.__version__}")

# TODO: ganti dengan nama file datasetmu (parquet disarankan)
DATA_FILE = "raw/dataset.parquet"
path = resolve_data_path(DATA_FILE)
print(f"Dataset: {path}")
```

## 1. Memuat Data (Lazy)

Gunakan `pl.scan_parquet` / `pl.scan_csv` supaya file tidak langsung dibaca ke memori.


```python
# TODO: gunakan scan_* sesuai format file
lf = pl.scan_parquet(path)

ringkasan_dataframe(lf)
```

## 2. Skema & Cuplikan Data


```python
# Skema kolom dan tipe data
print(lf.collect_schema())
```

```python
# Cuplikan 10 baris pertama
lf.head(10).collect()
```

## 3. Profiling dengan DuckDB SUMMARIZE

`SUMMARIZE` memberi statistik ringkas tiap kolom dalam satu perintah.


```python
df = lf.collect()

duckdb.sql("SUMMARIZE SELECT * FROM df").df()
```

## 4. Missing Value & Duplikat


```python
hitung_missing(df)
```

```python
jumlah_duplikat = df.height - df.n_unique()
print(f"Total baris    : {df.height:,}")
print(f"Baris duplikat : {jumlah_duplikat:,}")
```
